# ECG example
Load the bundled synthetic fixture with the public BioSignalsPlux reader and process its ECG signal directly.

```python
from pathlib import Path
import numpy as np
from multimodalphysiokit.io import read_biosignalsplux_hdf5
from multimodalphysiokit.processors import ECGProcessor
from multimodalphysiokit.utils import plot_signal_with_phases

data_file = Path("examples/sample_data/synthetic_multimodal_180s.hdf5")
recording = read_biosignalsplux_hdf5(data_file)
recording.signal_names()
```

```python
phase_intervals_seconds = np.array([[0.0, 90.0], [90.0, 180.0]])
recording.set_phases(phase_intervals_seconds, ["segment_1", "segment_2"])
plot_signal_with_phases(recording, "ecg", title="Raw ECG signal", show=True)
```

```python
ecg_signal = recording.get_signal("ecg")
(ecg_signal.n_samples, ecg_signal.sampling_frequency, ecg_signal.duration_seconds)
```

```python
processor = ECGProcessor(label_frequency_analysis=1, return_intermediates=True)
result = processor.process(ecg_signal)
```

# Respiration example
Load the bundled synthetic fixture and process its RIP respiration signal directly.

```python
from pathlib import Path
import numpy as np
from multimodalphysiokit.io import read_biosignalsplux_hdf5
from multimodalphysiokit.processors import RespirationProcessor
from multimodalphysiokit.utils import plot_signal_with_phases

data_file = Path("examples/sample_data/synthetic_multimodal_180s.hdf5")
recording = read_biosignalsplux_hdf5(data_file)
recording.signal_names()
```

```python
phase_intervals_seconds = np.array([[0.0, 90.0], [90.0, 180.0]])
recording.set_phases(phase_intervals_seconds, ["segment_1", "segment_2"])
plot_signal_with_phases(recording, "rip", title="Raw respiration signal", show=True)
```

```python
respiration_signal = recording.get_signal("rip")
(respiration_signal.n_samples, respiration_signal.sampling_frequency, respiration_signal.duration_seconds)
```

```python
result = RespirationProcessor().process(respiration_signal)
```

```python
{
    "feature_count": len(result.features),
    "original_sampling_frequency": result.metadata["original_sampling_frequency"],
    "processed_sampling_frequency": result.metadata["processed_sampling_frequency"],
    "selected_features": {
        name: result.features[name]
        for name in (
            "respiration_breath_rate_mean",
            "respiration_inspiratory_time_mean",
            "respiration_amplitude_mean",
        )
    },
}
```

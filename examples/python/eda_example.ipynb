# EDA example
Direct EDA processing. The processor requires the external **cvxEDA** dependency; see `docs/dependencies.md`.

```python
from pathlib import Path
import numpy as np
from multimodalphysiokit.io import read_biosignalsplux_hdf5
from multimodalphysiokit.processors import EDAProcessor
from multimodalphysiokit.utils import plot_signal_with_phases

data_file = Path("examples/sample_data/synthetic_multimodal_180s.hdf5")
recording = read_biosignalsplux_hdf5(data_file)
recording.signal_names()
```

```python
phase_intervals_seconds = np.array([[0.0, 90.0], [90.0, 180.0]])
recording.set_phases(phase_intervals_seconds, ["segment_1", "segment_2"])
plot_signal_with_phases(recording, "eda", title="Raw EDA signal", show=True)
```

```python
eda_signal = recording.get_signal("eda")
(eda_signal.n_samples, eda_signal.sampling_frequency, eda_signal.duration_seconds)
```

```python
processor = EDAProcessor()
result = processor.process(eda_signal)
```

```python
{
    "feature_count": len(result.features),
    "selected_features": {
        name: result.features[name]
        for name in ("eda_scl_mean", "eda_scl_std", "eda_scr_per_min")
    },
}
```

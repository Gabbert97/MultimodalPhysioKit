# Temperature example
Process temperature directly, then process two offline segments through the loaded `Recording`.

```python
from pathlib import Path
import numpy as np
from multimodalphysiokit.io import read_biosignalsplux_hdf5
from multimodalphysiokit.processors import TemperatureProcessor
from multimodalphysiokit.utils import plot_signal_with_phases

data_file = Path("examples/sample_data/synthetic_multimodal_180s.hdf5")
recording = read_biosignalsplux_hdf5(data_file)
recording.signal_names()
```

```python
phase_intervals_seconds = np.array([[0.0, 90.0], [90.0, 180.0]])
recording.set_phases(phase_intervals_seconds, ["segment_1", "segment_2"])
plot_signal_with_phases(recording, "temp", title="Raw temperature signal", show=True)
```

```python
temperature_signal = recording.get_signal("temp")
processor = TemperatureProcessor()
result = processor.process(temperature_signal)
{name: result.features[name] for name in ("temperature_mean", "temperature_delta", "temperature_linear_slope")}
```

```python
phase_results = recording.process_phases(processor, signal_names="temp")
```

```python
{
    label: {
        "feature_count": len(value.features),
        "temperature_mean": value.features["temperature_mean"],
    }
    for label, value in phase_results.items()
}
```
